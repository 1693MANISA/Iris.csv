### 6. Write a Python program to create a plot to get general Stastics Of Iris Data

### 7. Write a Python program to create a Bar plot to get the frequency of the three species of the Iris Data

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df_modified, x='Species')
plt.title("Frequency of Each Iris Species")
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()
```

### 8. Write a python program to create a Pie plot to get the frequency of the three species of the Iris data

```python
species_counts = df_modified['Species'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(species_counts, labels=species_counts.index, autopct='%1.1f%%', startangle=140)
plt.title("Distribution of Iris Species")
plt.axis('equal')
plt.show()
```

### 10. Write a Python program to create a scatter plot using sepal length and petal_width to separate the Species classes.

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_modified, x='SepalLengthCm', y='PetalWidthCm', hue='Species', palette='Set1')
plt.title("Scatter Plot (Sepal Length vs Petal Width)")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.legend(title='Species')
plt.show()
```

### 11. Use the clustering algorithm and plot the respective chart


```python
X = df_modified[['PetalLengthCm','PetalWidthCm']]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X_scaled)
clusters = kmeans.labels_
df_modified['Cluster'] = clusters

# Plotting clusters
plt.figure(figsize=(8, 6))
sns.scatterplot(x=X_scaled[:, 0], y=X_scaled[:, 1], hue=clusters, palette='Set1')
plt.title("K-Means Clustering (Petal Length vs Petal Width)")
plt.xlabel("Petal Length (standardized)")
plt.ylabel("Petal Width (standardized)")
plt.legend(title='Cluster')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x=X_scaled[:, 0], y=X_scaled[:, 1], hue=df_modified['Species'], palette='Set2')
plt.title("Actual Species Labels")
plt.xlabel("Petal Length (standardized)")
plt.ylabel("Petal Width (standardized)")
plt.legend(title='Species')
plt.show()
```
